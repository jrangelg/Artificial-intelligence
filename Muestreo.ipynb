```r
# Tamano de muestra
e<-c(rep(0.1,3),rep(0.2,3),rep(0.3,3),rep(0.4,3),rep(0.5,3))
a<-c(rep(c(0.05,0.01,0.005),5))
a
matriz_base<-data.frame(e,a,n=rep(0,15))
matriz_base
for(i in 1:15){
  matriz_base[i,3]<-(qnorm(matriz_base[i,2]/2)**2)/(matriz_base[i,1]**2)
}
matriz_base
#MAS
library ( survey )
data ( iris )
iris$fpc_mas <- 500
N<-iris$fpc_mas[1]
n<-length(iris$Sepal.Length)
w<-N/n
pi<-n/N
w
diseno_mas <- svydesign( id = ~1 , weights = NULL ,#strata =
                         , fpc = ~fpc_mas ,
                           data = iris )
View(iris)
mean<-svymean(~Sepal.Length,diseno_mas)
mean
confint(mean)
sd(iris$Sepal.Length)/sqrt(150)
t.test(iris$Sepal.Length)
#Estratificado
iris$fpc_est<-200
diseno_est <- svydesign( id = ~1 ,strata = ~Species, weights = NULL , fpc = ~fpc_est ,
                         data = iris )
table(iris$Species)
mean<-svymean(~Sepal.Length,diseno_est)
mean
confint(mean)

View(Titanic)
df_titanic <-as.data.frame(Titanic)
diseno_cong <- svydesign(
id = ~Class ,
weights = ~Freq ,
data = df_titanic
)
prop_survived <- svymean(~ Survived , diseno_cong)
print(prop_survived)
prop.test(sum(df_titanic$Survived == 'Yes'),length(df_titanic$Survived == 'Yes'))
confint(prop_survived)
sd(prop_survived)/3
prop_survived

# Parámetros iniciales de ejemplo
# N_h: Tamaño de la población en cada estrato
# S_h: Desviación estándar de la variable en cada estrato
# c_h: Costo por unidad muestreada en cada estrato
N_h <- c(1000, 2500, 1500)
S_h <- c(12, 8, 15)
c_h <- c(4, 9, 16)
n_total <- 200 # Tamaño de muestra total deseado

N <- sum(N_h)

# 1. Afijación Proporcional al Tamaño
# n_h = n * (N_h / N)
n_prop <- round(n_total * (N_h / N))
n_prop
# 2. Afijación de Neyman (considera tamaño y variabilidad S_h)
# n_h = n * (N_h * S_h) / sum(N_h * S_h)
num_neyman <- N_h * S_h
n_neyman <- round(n_total * (num_neyman / sum(num_neyman)))

# 3. Afijación Óptima (considerando costos c_h y variabilidad S_h)
# n_h = n * (N_h * S_h / sqrt(c_h)) / sum(N_h * S_h / sqrt(c_h))
num_optimo <- (N_h * S_h) / sqrt(c_h)
n_optimo <- round(n_total * (num_optimo / sum(num_optimo)))

# Consolidar resultados en una tabla
resultados <- data.frame(
  Estrato = 1:length(N_h),
  N_h = N_h,
  S_h = S_h,
  Costo_h = c_h,
  Proporcional = n_prop,
  Neyman = n_neyman,
  Optimo_Costos = n_optimo
)

print(resultados)

library(survey)

# Cargar el conjunto de datos estratificado por defecto del paquete survey
data(api)
# apistrat contiene una muestra estratificada de escuelas (por tipo: E, M, H)
# Variables clave: stype (estrato), pw (pesos de muestreo), fpc (corrección por población finita), api00 (variable de interés)

# 1. Definir el diseño de muestreo estratificado
# id=~1: Muestreo aleatorio simple dentro de estratos
# strata=~stype: Variable de estratificación
# weights=~pw: Pesos muestrales (inversos de probabilidad de inclusión)
# fpc=~fpc: Corrección para población finita (tamaño poblacional por estrato)
dstrat <- svydesign(id = ~1, strata = ~stype, weights = ~pw, data = apistrat, fpc = ~fpc)
View(apistrat)
sum(apistrat$pw)
length(apistrat$pw)
# --- EJEMPLO 1: Estimación de la Media Poblacional ---
# Estimar la media del puntaje académico en el año 2000 (api00)
media_api00 <- svymean(~api00, dstrat)
print("--- Estimación de la Media ---")
print(media_api00)
# Intervalo de confianza al 95% para la media
confint(media_api00)
apistrat$fpcmas<-1000
dsmas <- svydesign(id = ~1, data = apistrat, fpc = ~fpcmas)
mediamas<-svymean(~api00,dsmas)
confint(mediamas)
# --- EJEMPLO 2: Estimación del Total Poblacional ---
# Estimar el total de la matrícula estudiantil (enroll) o del puntaje acumulado
total_enroll <- svytotal(~enroll, dstrat)
print("--- Estimación del Total ---")
print(total_enroll)
confint(total_enroll)


# --- EJEMPLO 3: Estimación por Dominios (Subpoblaciones) ---
# Estimar la media y el total de api00 desglosado por un dominio (ej. si la escuela recibe premios 'awards')
media_por_dominio <- svyby(~api00, ~awards, dstrat, svymean)
total_por_dominio <- svyby(~api00, ~awards, dstrat, svytotal)

print("--- Media por Dominio (awards) ---")
print(media_por_dominio)

print("--- Total por Dominio (awards) ---")
print(total_por_dominio)
```
